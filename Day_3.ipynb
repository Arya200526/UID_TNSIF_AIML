```python
n = int(input("Enter a number: "))
is_prime = True

if n <= 2:
    is_prime = False
else:
    for i in range(2, n):
        if n % i == 0:
            is_prime = False
            break

if is_prime:
    print("Prime number")
else:
    print("Not a prime number")
```

```python
def palindrome():
    n = int(input("Enter a number: "))
    num = n
    result = 0

    while num > 0:
        ld = num % 10
        result = result * 10 + ld
        num = num // 10

    if n == result:
        print("Palindrome")
    else:
        print("Not palindrome")


palindrome()
```

```python
class Recatangle():

  def __init__(self,length,width):
    self.length = length
    self.width  = width

  def area(self):
    return self.length * self.width

  def perimeter(self):
    return 2 *(self.length + self.width)

r1 = Recatangle(8,9)
print(r1.area())
print(r1.perimeter())
```
